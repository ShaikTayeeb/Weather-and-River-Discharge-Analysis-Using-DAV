MATPLOTLIB

```python
plt.figure(figsize=(12, 5))

plt.plot(
    df['date'],
    df['temperature_mean_c']
)

plt.xlabel("Date")
plt.ylabel("Temperature (°C)")
plt.title("Temperature Variation Over Time")

plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    df['date'],
    df['river_discharge_m3s']
)

plt.xlabel("Date")
plt.ylabel("River Discharge (m³/s)")
plt.title("River Discharge Over Time")

plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    df['date'],
    df['rain_mm']
)

plt.xlabel("Date")
plt.ylabel("Rainfall (mm)")
plt.title("Rainfall Over Time")

plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df['temperature_mean_c'], bins=30)

plt.xlabel("Temperature (°C)")
plt.ylabel("Frequency")
plt.title("Distribution of Temperature")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df['river_discharge_m3s'], bins=30)

plt.xlabel("River Discharge (m³/s)")
plt.ylabel("Frequency")
plt.title("Distribution of River Discharge")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df['temperature_mean_c'],
    df['river_discharge_m3s']
)

plt.xlabel("Temperature (°C)")
plt.ylabel("River Discharge (m³/s)")
plt.title("Temperature vs River Discharge")

plt.show()
```

SEABORN

```python
sns.set_style("whitegrid")
```

```python
plt.figure(figsize=(12, 8))

correlation = df.select_dtypes(
    include='number'
).corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    x=df['temperature_mean_c']
)

plt.title("Temperature Distribution")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    df['rain_mm'],
    bins=30,
    kde=True
)

plt.xlabel("Rainfall (mm)")
plt.title("Rainfall Distribution")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='temperature_mean_c',
    y='river_discharge_m3s'
)

plt.title("Temperature vs River Discharge")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='rain_mm',
    y='river_discharge_m3s'
)

plt.xlabel("Rainfall (mm)")
plt.ylabel("River Discharge (m³/s)")
plt.title("Rainfall vs River Discharge")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.regplot(
    data=df,
    x='rain_mm',
    y='river_discharge_m3s'
)

plt.title("Rainfall and River Discharge Relationship")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='temperature_mean_c',
    y='relative_humidity_mean_pct'
)

plt.xlabel("Temperature (°C)")
plt.ylabel("Relative Humidity (%)")
plt.title("Temperature vs Relative Humidity")

plt.show()
```
